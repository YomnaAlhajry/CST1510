```python
S["percent"] = (S["used"] / S["total"]) * 100
over_limit = S[S["percent"] >= 100]

print(over_limit)
print(S)
```

```python
def classify(percent):
    if percent >= 100:
        return "Over Limit"
    elif percent >= 90:
        return "WARNING"
    else:
        return "OK"
    
S["status"] = S["percent"].apply(classify)
print(S)
```
